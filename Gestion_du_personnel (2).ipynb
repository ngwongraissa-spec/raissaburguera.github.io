### 1.1 Demande de la saisie de la civilité :
- Si la saisie est correcte, le programme passe au champ suivant.
- Si la saisie est incorrecte, un message apparait avec les consignes de saisie et la saisie de la civilité est redemandée.

```python
civilité=str(input("Saisir M. ou Mme : "))

if civilité=="M." or civilité=="Mme" :
    civilité=civilité

else :
    print("Attention! Saisir M. ou Mme")
    civilité=str(input("Saisir M. ou Mme : "))
```

### 1.2 Demande de la saisie du Nom :
- Chaque lettre du Nom saisi est comparée aux caractères autorisés dans la variable "alphabet".
- Quand la comparaison de l'ensemble des caractères du Nom n'est pas fausse, le programme passe au champ suivant.
- Dans la boucle for, si un des caractères ne fait pas parti de ceux autorisés dans la variable "alphabet", un message apparait avec les consignes de saisie et la saisie du Nom est redemandée. La boucle while recommense jusqu'à ce que le Nom soit conforme.
- break permet d'arrêter la boucle for au 1er caractère non conforme.Cela optimise la durée d'exécution.

```python
alphabet = ("abcdefghijklmnopqrstuvwxyz-'éèçAZERTYUIOPQSDFGHJKLMWXCVBN")
validation_nom = False
while not validation_nom:
    Nom = input('saisir le nom : ')
    validation_nom = True

    for lettre in Nom:
        if lettre not in alphabet:
            validation_nom = False
            print("ATTENTION! caractères spéciaux et chiffres non autorisés")
            break
```

### 1.3 Demande de la saisie du Prénom :
Idem que pour le Nom

```python
alphabet = ("abcdefghijklmnopqrstuvwxyz-'éèçAZERTYUIOPQSDFGHJKLMWXCVBN")
validation_prenom = False
while not validation_prenom :
    Prénom = input('saisir le prénom : ')
    validation_prenom = True

    for lettre in Prénom:
        if lettre not in alphabet:
            validation_prenom = False
            print("ATTENTION! caractères spéciaux et chiffres non autorisés")
            break
```

### 3.6 Pourcentage de noms composés

- La variable "compteur" compte le nombre de caractères "-" de chaque ligne pour la colonne "Nom".
